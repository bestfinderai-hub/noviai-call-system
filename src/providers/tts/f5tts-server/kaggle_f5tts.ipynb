```python
# Cell 4 â€” Starta server (laddar modell ~2-3 min)
import subprocess, sys, time, urllib.request, os

print('[4/5] Startar F5-TTS-server (laddar 3.4 GB modell, ~2-3 min)...')

env = os.environ.copy()
env['MODEL_DIR'] = '/kaggle/working/model'
env['REF_AUDIO']  = '/kaggle/working/reference.wav'
env['REF_TEXT']   = 'Kom afton, sÃ¤nk dig neder, du ger oss mÃ¶dans lÃ¶n. Vad lugn du oss bereder, natur vad du Ã¤r skÃ¶n. Se aftonrÃ¶dnan vilar i dalens vala famn. FrÃ¥n fÃ¤stet solen ilar bort till en annan hand.'

srv = subprocess.Popen(
    [sys.executable, '-m', 'uvicorn', 'f5tts_server:app',
     '--host', '0.0.0.0', '--port', '8000'],
    cwd='/kaggle/working',
    stdout=open('/kaggle/working/server.log', 'w'),
    stderr=subprocess.STDOUT,
    env=env,
    start_new_session=True
)

print('VÃ¤ntar pÃ¥ att modell laddas...')
for i in range(120):
    time.sleep(3)
    try:
        urllib.request.urlopen('http://localhost:8000/health', timeout=2)
        print(f'Server OK! ({i*3}s)')
        break
    except:
        if i % 10 == 0:
            print(f'  [{i*3}s] laddar modell...')
else:
    print('TIMEOUT â€” server.log:')
    print(open('/kaggle/working/server.log').read()[-3000:])
    raise SystemExit(1)

print('Cell 4 KLAR')
```

```python
# Cell 5 â€” Cloudflare-tunnel + spara URL till HuggingFace + keepalive
import subprocess, re, os, time, urllib.request, io

HF_TOKEN = 'YOUR_HF_TOKEN_HERE'

print('[5/5] Laddar cloudflared...')
subprocess.run([
    'wget', '-q',
    'https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64',
    '-O', '/kaggle/working/cloudflared'
], check=True)
os.chmod('/kaggle/working/cloudflared', 0o755)

print('[5/5] Startar tunnel...')
tunnel_log = open('/kaggle/working/tunnel.log', 'w')
tun = subprocess.Popen(
    ['/kaggle/working/cloudflared', 'tunnel', '--url', 'http://localhost:8000'],
    stdout=tunnel_log, stderr=subprocess.STDOUT,
    start_new_session=True
)

tunnel_url = None
for i in range(60):
    time.sleep(2)
    try:
        log = open('/kaggle/working/tunnel.log').read()
        urls = re.findall(r'https://[\w-]+\.trycloudflare\.com', log)
        if urls:
            tunnel_url = urls[0]
            break
    except:
        pass

if not tunnel_url:
    print('TIMEOUT â€” tunnel.log:')
    print(open('/kaggle/working/tunnel.log').read())
    raise SystemExit(1)

print(f'\n======================================================')
print(f'  F5TTS_SERVER_URL={tunnel_url}')
print(f'======================================================')

# Spara URL till HuggingFace â€” NovAI kan hÃ¤mta den automatiskt
try:
    from huggingface_hub import upload_file
    upload_file(
        path_or_fileobj=io.BytesIO(tunnel_url.encode()),
        path_in_repo='tunnel_url.txt',
        repo_id='bestfinderai/f5tts-swedish-male',
        token=HF_TOKEN,
        commit_message='tunnel url update'
    )
    print('URL sparad till HuggingFace âœ“')
    print(f'HÃ¤mta med: https://huggingface.co/bestfinderai/f5tts-swedish-male/resolve/main/tunnel_url.txt')
except Exception as e:
    print(f'HF upload misslyckades: {e}')

open('/kaggle/working/tunnel_url.txt', 'w').write(tunnel_url)

# Keepalive â€” fÃ¶rhindrar Kaggles 40-min idle-timeout
print('\nKeepalive aktiv â€” pingar server var 30s...')
tick = 0
while tun.poll() is None:
    time.sleep(30)
    tick += 1
    try:
        urllib.request.urlopen('http://localhost:8000/health', timeout=5)
        if tick % 20 == 0:
            print(f'[{tick*30//60}min] OK â€” {tunnel_url}')
    except Exception as e:
        print(f'[{tick*30}s] VARNING: {e}')

print('Tunnel avslutad.')
```
