```python
# Cell 4 â€” Starta server som subprocess (vÃ¤ntar pÃ¥ Server OK innan tunnel)
import subprocess, sys, time, urllib.request, os

print('[4/5] Startar TTS-server (laddar modell ~2-3 min)...')

env = os.environ.copy()
env['REF_AUDIO'] = '/kaggle/working/reference.wav'
env['DEVICE'] = 'cuda'

srv = subprocess.Popen(
    [sys.executable, '-m', 'uvicorn', 'cb_server:app', '--host', '0.0.0.0', '--port', '8000'],
    cwd='/kaggle/working',
    stdout=open('/kaggle/working/server.log','w'),
    stderr=subprocess.STDOUT,
    env=env
)

for i in range(120):
    time.sleep(3)
    try:
        urllib.request.urlopen('http://localhost:8000/health', timeout=2)
        print('Server OK!')
        break
    except:
        if i % 10 == 0:
            print(f'  [{i*3}s] laddar modell...')
else:
    print('TIMEOUT â€” server-log:')
    print(open('/kaggle/working/server.log').read()[-3000:])
    raise SystemExit(1)

print('Cell 4 KLAR')
```

```python
# Cell 5 â€” Starta Cloudflare-tunnel + keepalive (hÃ¥ller sessionen vid liv)
# KRITISKT: Skriv till LOG-FIL (inte PIPE) â€” annars blockerar cloudflared nÃ¤r buffern fylls
import subprocess, re, os, time, urllib.request

print('[5/5] Laddar ned cloudflared...')
subprocess.run([
    'wget','-q',
    'https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64',
    '-O','/kaggle/working/cloudflared'
], check=True)
os.chmod('/kaggle/working/cloudflared', 0o755)

print('[5/5] Startar tunnel (output â†’ tunnel.log)...')
tunnel_log = open('/kaggle/working/tunnel.log', 'w')
tun = subprocess.Popen(
    ['/kaggle/working/cloudflared','tunnel','--url','http://localhost:8000'],
    stdout=tunnel_log, stderr=subprocess.STDOUT
)

# LÃ¤s URL frÃ¥n log-filen (inte frÃ¥n PIPE)
tunnel_url = None
for i in range(60):
    time.sleep(2)
    try:
        log = open('/kaggle/working/tunnel.log').read()
        urls = re.findall(r'https://[\w-]+\.trycloudflare\.com', log)
        if urls:
            tunnel_url = urls[0]
            break
    except:
        pass
    if i % 5 == 0:
        print(f'  [{i*2}s] VÃ¤ntar pÃ¥ tunnel...')

if tunnel_url:
    print(f'\n======================================================')
    print(f'  CHATTERBOX_SERVER_URL={tunnel_url}')
    print(f'======================================================')
    print('Kopiera URL:en ovan till NovAI .env och sÃ¤tt TTS_PROVIDER=chatterbox')
else:
    print('TIMEOUT â€” tunnel.log:')
    print(open('/kaggle/working/tunnel.log').read())
    raise SystemExit(1)

# Keepalive: pingas server var 30s â€” fÃ¶rhindrar Kaggles 40-min idle-timeout
print('\nKeepalive aktiv â€” pingar server var 30s, skriver status var 10 min...')
tick = 0
while tun.poll() is None:
    time.sleep(30)
    tick += 1
    try:
        urllib.request.urlopen('http://localhost:8000/health', timeout=5)
        if tick % 20 == 0:
            print(f'[{tick*30//60}min] Server OK â€” {tunnel_url}')
    except Exception as e:
        print(f'[{tick*30}s] VARNING: server svarar inte: {e}')

print('Tunneln avslutad.')
```
